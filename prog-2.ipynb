Asian time Zone

```python
import datetime
import pytz
ist=pytz.timezone("Asia/Kolkata")
curr_date_time=datetime.datetime.now(ist)
print(f"current date and time is {curr_date_time}")
print("Full month name: ",curr_date_time.strftime("%B"))
print("abb month name: ",curr_date_time.strftime("%b"))
print("Full weekday name: ",curr_date_time.strftime("%A"))
print("abb weekday name: ",curr_date_time.strftime("%a"))
print("2 digit year",curr_date_time.strftime("%y"))
print("date is only day ",curr_date_time.strftime("%d"))
print("date is ",curr_date_time.strftime("%D"))
print("time is (24hrs)",curr_date_time.strftime("%H"))
print("time is (12hrs)",curr_date_time.strftime("%I"))
print("Minutes: ",curr_date_time.strftime("%M"))
print("Seconds: ",curr_date_time.strftime("%S"))
print("AM/PM: ",curr_date_time.strftime("%p"))
```
