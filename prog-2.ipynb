Reverse a list using slice method

```python
def reverse_list(list):
    return list[::-1]
list=[1,2,3,4,5]
print("Original list",list)

print("After slicing list",list[2:5:1])
list=reverse_list(list)
print("Reverse a list using slicing",list)
```

reverse a list with rever method

```python
def reverse_list(list):
    list.reverse()
    return list
list=[1,2,3,4,5,6]
print("Original list",list)
list=reverse_list(list)
print("Reverse a list using slicing",list)
```
