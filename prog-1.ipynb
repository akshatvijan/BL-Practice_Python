```python
import random

def run_flip_a_coin(n):
    cnt=0
    for _ in range(n):
        list_choice=["Head","Tail"]
        ans=random.choice(list_choice)
        print(ans)
        if ans=="Head":
            cnt+=cnt
    percentage=(cnt/n)
    print(f"NUmber of time head is {cnt} and percentage is {percentage:.2f}")

num_flip=int(input("Enter the number of coin flips: "))
run_flip_a_coin(num_flip)
```
