Prog-1.1 sum of list

```python
def custom_sum(list):
    total=0
    for i in list:
        total+=i
    return total
list=[10,28,43,656,8]
custom_result=custom_sum(list)
print("Custom result is ",custom_result)
builtin_result=sum(list)
print("Built in result is ",builtin_result)
print(builtin_result==custom_result,"for custom result and built-in result")
```

1.2 sum of list

```python
def custom_sum(list):
    total=0
    for i in list:
        total+=i
    return total
list=[]
size=int(input("Enter the size of a list"))
for _ in range(size):
    num=int(input("Enter elements os a list"))
    list.append(num)

custom_result=custom_sum(list)
print("Custom result is ",custom_result)
builtin_result=sum(list)
print("Built in result is ",builtin_result)
print(builtin_result==custom_result,"for custom result and built-in result")
```
