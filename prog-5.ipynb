```python
def fact(n:int):
    
    if n==0:
        return 1
    else:
        result=1
        for i in range(1,n+1):
            result=result*i
        return result
    


n=int(input("Enter the number"))
print(fact(n))
```

```python
def fact(n:int):
    
    if n==0:
        return 1
    
    return n*fact(n-1)

n=int(input("Enter the number"))
print(fact(n))
```
