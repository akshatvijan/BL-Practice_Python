list indexing

```python
list=[10,20,30,40,50,60]
print(list[2])
print(list[-3])
# print(list[10]) IndexError
```

slicing in list

```python
list=[1,2,3,4,5,6,7]
print(list[1:4])
print(list[:4])
print(list[3:])
print(list[0:7:2])
print(list[::-1])
```

traversing a list

```python
list=[1,2,3,4,5]
for i in list:
    print(i)

for i in range(len(list)):
    print("element at index",i," is ",list[i])
```

enumerate

```python
list = [1,2,3,4,5]
for index,value in enumerate(list):
    print(index,' ',value)
```

```python
list=[43,23,48,32,7,3,5]
list.sort()
print(list) 
a=list.sorted()
```
