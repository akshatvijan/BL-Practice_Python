PROG 1: Calculator App

```python
symbol=input("enter the symbol for computation (+,-,*,/,%)")
a=int(input("Enter the first number"))
b=int(input("Enter the second number"))

if symbol=="+":
    print(a+b)
elif symbol=="-":
    print(a-b)
elif symbol=="*":
    print(a*b)
elif symbol=="/":
    print(a/b)
elif symbol=="%":
    print(a%b)
```

PROG 2 : Multiplies Element In A List

```python
list=[]
num=int(input("Enter the size of a list"))
for i in range(num):
    a=int(input("Enter the elements of a list"))
    list.append(a)
product=1
for i in list:
    product=product*i
print(f"{product}")
```

PROG 3: Check Prime

```python
for num in range(1,101):
    if num<2:
        continue
    flag=True
    for i in range(2,num):
        if num%i==0:
            flag=False
            break
    if flag==True:
        print(f"{num}",end=' ')
```

PROG 5: Count Down using Recursion

```python
def countdown(num:int):
    if(num<0):
        return
    print(num)
    countdown(num-1)
    
    
num=int(input("Enter the number"))
countdown(num)
```
